```python
from pathlib import Path
import tempfile
import shutil
import kagglehub

handle = "saifzaman123445/daicwoz"
final_dir = Path("../data/raw/daicwoz")
final_dir.mkdir(parents=True, exist_ok=True)

with tempfile.TemporaryDirectory() as tmpdir:
    temp_dir = Path(tmpdir)

    for i in range(300, 493):
        rel_path = f"daicwoz/daicwoz/{i}_TRANSCRIPT.csv"
        try:
            downloaded = Path(
                kagglehub.dataset_download(
                    handle,
                    path=rel_path,
                    output_dir=str(temp_dir),
                )
            )

            src = downloaded if downloaded.is_file() else next(downloaded.rglob("*_TRANSCRIPT.csv"))
            dest = final_dir / src.name

            if dest.exists():
                dest.unlink()

            shutil.move(str(src), str(dest))
            print(f"Downloaded {dest.name}")
        except Exception as e:
            print(f"Skipped {i}: {e}")
```
